## customer id discrepancy 

```python
cols = [
    'customer_name',
    'gender',
    'age',
    'country',
    'city',
    'account_creation_date'
]

for customer_id, group in df_final.groupby('customer_id'):
    
    inconsistent_cols = []
    
    for col in cols:
        if group[col].nunique(dropna=False) > 1:
            inconsistent_cols.append(col)
    
    if inconsistent_cols:
        print(f"\nCustomer ID: {customer_id}")
        print("Inconsistent columns:", inconsistent_cols)
        
        for col in inconsistent_cols:
            print(f"{col}: {group[col].unique()}")
```

```python
cols = [
    'customer_name',
    'gender',
    'age',
    'country',
    'city',
    'account_creation_date'
]

check = df_final.groupby('customer_id')[cols].nunique(dropna=False)

bad_ids = check[(check > 1).any(axis=1)].index

print(bad_ids)
```

## Creating new columns

```python
warehouse_dict = {warehouse: i + 1 for i, warehouse in enumerate(sorted(df1['warehouse_location'].dropna().unique()))}
country_dict = {country: i + 1 for i, country in enumerate(sorted(df1['country'].dropna().unique()))}
city_dict = {city: i + 1 for i, city in enumerate(sorted(df1['city'].dropna().unique()))}
```

```python
df1['warehouse_location_code'] = (df1['warehouse_location'].map(warehouse_dict))
df1['country_code'] = (df1['country'].map(country_dict))
df1['city_code'] = (df1['city'].map(city_dict))
```

```python
df1['warehouse_id'] = (df1['warehouse_location_code'].astype(str)+ df1['country_code'].astype(str).str.zfill(2)+ df1['city_code'].astype(str).str.zfill(2))
df1.drop(columns=['warehouse_location_code','country_code','city_code','expected_country'],inplace=True)
df1['warehouse_id'] = df1['warehouse_id'].astype(int)
```

```python
category_dict = {category: i + 1 for i, category in enumerate(sorted(df1['category'].dropna().unique()))}
subcategory_dict = {subcategory: i + 1 for i, subcategory in enumerate(sorted(df1['sub_category'].dropna().unique()))}
```

```python
df1['category_code'] = (df1['category'].map(category_dict))
df1['subcategory_code'] = (df1['sub_category'].map(subcategory_dict))
```

```python
df1['inventory_id'] = (df1['category_code'].astype(str).str.zfill(2)+ df1['subcategory_code'].astype(str).str.zfill(2))
df1['inventory_id']=df1['inventory_id'].astype(int)
df1.drop(columns=['category_code','subcategory_code'],inplace=True)
```

```python
df1['order_date'] = pd.to_datetime(
    df1[['order_year', 'order_month', 'order_day']].rename(
        columns={
            'order_year': 'year',
            'order_month': 'month',
            'order_day': 'day'
        }
    )
)
```
