```python
import os
import time
import tempfile
from pathlib import Path
from datetime import date, datetime
from decimal import Decimal, ROUND_HALF_UP
from dotenv import load_dotenv


import pandas as pd
from sqlalchemy import create_engine, inspect, text
from sqlalchemy.engine import URL
from sqlalchemy.exc import OperationalError, InterfaceError
from sqlalchemy.pool import NullPool
load_dotenv("../.env")

# ============================================================
# 1. SETTINGS — TEAMMATE'S PC VIA TAILSCALE
# ============================================================
DB_HOST = os.getenv("DB_HOST")
DB_PORT = int(os.getenv("DB_PORT", 3306))
DB_USER = os.getenv("DB_USER")
DB_PASSWORD = os.getenv("DB_PASSWORD")
DB_NAME = os.getenv("DB_NAME")

# True = ALL transaction dates in df_final. False = old selected interval.
UPLOAD_ALL_DATES = False

START_DATE = "2024-02-03"
END_DATE = "2025-08-03"



DRY_RUN = False       # First run True; then False to actually INSERT.
MAX_NEW_ROWS_PER_RUN = None  # None = upload all remaining rows in every table.
PREFER_BULK = True   # LOAD DATA LOCAL INFILE if server permits; else SQL INSERT.
BULK_ROWS = 10_000
BULK_BYTES = 2_000_000
SQL_ROWS = 500
MAX_RETRIES = 5
PROGRESS_EVERY = 10_000

# ============================================================
# 2. SEVEN EXISTING TABLES AND THEIR COLUMNS
# ============================================================
TABLE_COLUMNS = {
    "customers": ["customer_id", "customer_name", "gender", "age",
                  "customer_segment", "country", "city",
                  "customer_loyalty_score", "account_creation_date"],
    "product": ["product_id", "product_name", "category", "sub_category",
                "product_rating_avg", "product_reviews_count", "unit_price_usd"],
    "warehouse": ["warehouse_id", "warehouse_location"],
    "orders": ["order_id", "order_date", "is_weekend", "order_status",
               "payment_method", "payment_status", "installment_plan",
               "device_type", "customer_id"],
    "inventory": ["inventory_id", "product_id", "warehouse_id", "stock_quantity"],
    "order_items": ["order_id", "product_id", "warehouse_id", "quantity",
                    "discount_percent", "total_price_usd", "cost_usd",
                    "profit_usd", "tax_usd"],
    "shipment": ["order_id", "shipping_method", "shipping_cost_usd",
                 "delivery_days", "shipping_country", "delivery_status"],
}
PRIMARY_KEYS = {
    "customers": ["customer_id"],
    "product": ["product_id"],
    "warehouse": ["warehouse_id"],
    "orders": ["order_id"],
    "inventory": ["inventory_id"],
    "order_items": ["order_id", "product_id", "warehouse_id"],
    "shipment": ["order_id"],
}
LOAD_ORDER = list(TABLE_COLUMNS)
REQUIRED = {
    "customers": ["customer_id", "customer_name"],
    "product": ["product_id", "product_name"],
    "warehouse": ["warehouse_id", "warehouse_location"],
    "orders": ["order_id", "order_date", "is_weekend", "customer_id"],
    "inventory": ["inventory_id", "product_id", "warehouse_id", "stock_quantity"],
    "order_items": ["order_id", "product_id", "warehouse_id", "quantity"],
    "shipment": ["order_id"],
}
MONEY_COLUMNS = {
    "customers": ["customer_loyalty_score"],
    "product": ["product_rating_avg", "unit_price_usd"],
    "order_items": ["discount_percent", "total_price_usd", "cost_usd", "profit_usd", "tax_usd"],
    "shipment": ["shipping_cost_usd"],
}
INT_COLUMNS = {"age", "product_id", "warehouse_id", "inventory_id",
               "product_reviews_count", "stock_quantity", "quantity",
               "delivery_days", "is_weekend"}
ENUMS = {
    "gender": {"Male", "Female"},
    "customer_segment": {"Regular", "VIP", "Premium"},
    "order_status": {"Completed", "Returned", "Pending", "Cancelled", "Processing"},
    "payment_method": {"Apple Pay", "Bank Transfer", "Credit Card", "PayPal", "Debit Card"},
    "payment_status": {"Paid", "Pending", "Failed"},
    "installment_plan": {"Yes", "No"},
    "device_type": {"Mobile", "Desktop", "Tablet"},
    "shipping_method": {"Economy", "Next Day", "Express", "Standard"},
    "delivery_status": {"In Transit", "Delivered", "Failed", "Pending"},
}
BOUNDS = {"age": (14, 120), "customer_loyalty_score": (0, 100),
          "product_rating_avg": (0, 5), "product_reviews_count": (0, None),
          "unit_price_usd": (0, None), "stock_quantity": (0, None),
          "quantity": (1, None), "discount_percent": (0, 100),
          "total_price_usd": (0, None), "cost_usd": (0, None),
          "tax_usd": (0, None), "shipping_cost_usd": (0, None),
          "delivery_days": (0, None)}

# ============================================================
# 3. VALIDATED DATA PREPARATION (IN THE SAME CELL)
# ============================================================
if "df_final" not in globals() or df_final.empty:
    raise RuntimeError("df_final missing or empty. Run your upstream cleaning cell first.")

def order_timestamps(frame):
    dt = pd.to_datetime(frame["order_date"], errors="raise")
    if dt.isna().any():
        raise ValueError("order_date contains NULL")
    parts = ["order_hour", "order_minute", "order_second"]
    if not all(c in frame for c in parts):
        return dt
    nums = []
    for col, upper in zip(parts, (23, 59, 59)):
        x = pd.to_numeric(frame[col], errors="coerce")
        if x.isna().any() or not x.between(0, upper).all() or not x.mod(1).eq(0).all():
            raise ValueError(f"Invalid {col}")
        nums.append(x.astype("int64"))
    composed = (dt.dt.normalize()
                + pd.to_timedelta(nums[0], unit="h")
                + pd.to_timedelta(nums[1], unit="m")
                + pd.to_timedelta(nums[2], unit="s"))
    nonmidnight = dt.ne(dt.dt.normalize())
    if (nonmidnight & dt.dt.floor("s").ne(composed)).any():
        raise ValueError("order_date disagrees with hour/minute/second")
    return composed

def as_money(value):
    if pd.isna(value):
        return None
    number = Decimal(str(value))
    if not number.is_finite():
        raise ValueError(f"Nonfinite monetary value: {value}")
    return format(number.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP), ".2f")

def as_integer(value):
    if pd.isna(value):
        return None
    number = Decimal(str(value))
    if not number.is_finite() or number != number.to_integral_value():
        raise ValueError(f"Expected an integer, found {value}")
    return int(number)

source = df_final
if UPLOAD_ALL_DATES:
    selected = source.copy()
    print(f"Source selection: ALL dates, {len(selected):,} transaction rows")
else:
    dates = pd.to_datetime(source["order_date"], errors="raise")
    start, end = pd.Timestamp(START_DATE), pd.Timestamp(END_DATE)
    if start > end:
        raise ValueError("START_DATE is after END_DATE")
    selected = source.loc[dates.ge(start) & dates.lt(end + pd.Timedelta(days=1))].copy()
    print(f"Source selection: {START_DATE} to {END_DATE}, {len(selected):,} rows")
if selected.empty:
    raise ValueError("No transactions selected")

# Transaction timestamps, weekend flag, customer account date
selected["order_date"] = order_timestamps(selected)
selected["account_creation_date"] = pd.to_datetime(
    selected["account_creation_date"], errors="raise"
).dt.date
week_map = {"true": 1, "false": 0, "yes": 1, "no": 0,
            "1": 1, "0": 0, "weekend": 1, "weekday": 0}
selected["is_weekend"] = (selected["is_weekend"].astype("string")
                          .str.strip().str.lower().map(week_map))
if selected["is_weekend"].isna().any():
    raise ValueError("Unrecognized is_weekend values")

# Product identities and representative median attributes
product_identity = selected.groupby("product_id")[["product_name", "category", "sub_category"]].nunique(dropna=False)
if product_identity.gt(1).any().any():
    raise ValueError("Product name/category/subcategory conflicts")
products = (selected.groupby("product_id", as_index=False)
            .agg(product_name=("product_name", "first"),
                 category=("category", "first"),
                 sub_category=("sub_category", "first"),
                 product_rating_avg=("product_rating_avg", "median"),
                 product_reviews_count=("product_reviews_count", "median"),
                 unit_price_usd=("unit_price_usd", "median")))
products["product_reviews_count"] = products["product_reviews_count"].round()

# Inventory IDs based on ALL original product/warehouse pairs;
# averaged stock per pair based on SELECTED period (not latest stock).
pair_cols = ["product_id", "warehouse_id"]
all_pairs = source[pair_cols].drop_duplicates().sort_values(pair_cols).reset_index(drop=True)
if all_pairs.isna().any().any():
    raise ValueError("Missing product/warehouse IDs")
all_pairs["inventory_id"] = range(1, len(all_pairs) + 1)
inventory = (selected.groupby(pair_cols, as_index=False)
             .agg(stock_quantity=("stock_quantity", "mean")))
if inventory["stock_quantity"].isna().any():
    raise ValueError("NULL mean stock_quantity")
inventory["stock_quantity"] = inventory["stock_quantity"].round()
inventory = inventory.merge(all_pairs, on=pair_cols, validate="one_to_one")

# Most recent customer profile within selected data
latest_customer = selected.groupby("customer_id")["order_date"].transform("max")
customers = selected.loc[selected["order_date"].eq(latest_customer), TABLE_COLUMNS["customers"]].drop_duplicates()

prepared = {}
for table in LOAD_ORDER:
    cols = TABLE_COLUMNS[table]
    if table == "customers":
        frame = customers[cols].copy()
    elif table == "product":
        frame = products[cols].copy()
    elif table == "inventory":
        frame = inventory[cols].copy()
    else:
        missing = set(cols) - set(selected.columns)
        if missing:
            raise ValueError(f"Missing source columns for {table}: {sorted(missing)}")
        frame = selected[cols].drop_duplicates().copy()
    if frame[REQUIRED[table]].isna().any().any():
        raise ValueError(f"NULL in required columns of {table}")
    if frame.duplicated(PRIMARY_KEYS[table]).any():
        raise ValueError(f"Duplicate/conflicting primary keys in {table}")
    if table == "inventory" and frame.duplicated(pair_cols).any():
        raise ValueError("Duplicate inventory product/warehouse pair")

    for c in (INT_COLUMNS & set(frame.columns)):
        frame[c] = frame[c].map(as_integer)
    for c in MONEY_COLUMNS.get(table, []):
        frame[c] = frame[c].map(as_money)
    for c, allowed in ENUMS.items():
        if c in frame.columns:
            invalid = set(frame[c].dropna().unique()) - allowed
            if invalid:
                raise ValueError(f"{table}.{c}: invalid values {invalid}")
    for c, (lo, hi) in BOUNDS.items():
        if c in frame.columns:
            values = pd.to_numeric(frame[c].dropna())
            if values.lt(lo).any() or (hi is not None and values.gt(hi).any()):
                raise ValueError(f"{table}.{c}: violates expected numeric range")

    prepared[table] = frame.sort_values(PRIMARY_KEYS[table], kind="stable").reset_index(drop=True)
    print(f"Prepared {table:12s}: {len(prepared[table]):>10,} records")

print(f"Total expected SQL rows: {sum(len(v) for v in prepared.values()):,}")

# ============================================================
# 4. NEW DATABASE CONNECTION + SCHEMA PREFLIGHT
# ============================================================
engine = create_engine(
    URL.create(
        "mysql+pymysql",
        username=DB_USER,
        password=DB_PASSWORD,
        host=DB_HOST,
        port=DB_PORT,
        database=DB_NAME,
    ),
    poolclass=NullPool,
    connect_args={
        "local_infile": True,
        "connect_timeout": 15,
        "read_timeout": 180,
        "write_timeout": 180,
        "ssl_disabled": True,
    },
)

def check_destination():
    with engine.connect() as conn:
        actual_db = conn.execute(text("SELECT DATABASE() ")).scalar_one()
        if actual_db != DB_NAME:
            raise RuntimeError(f"Connected to unexpected database {actual_db!r}")
        inspector = inspect(conn)
        existing_tables = set(inspector.get_table_names())
        missing_tables = set(LOAD_ORDER) - existing_tables
        if missing_tables:
            raise RuntimeError(f"Teammate has not created tables: {sorted(missing_tables)}")
        for table in LOAD_ORDER:
            database_columns = {x["name"] for x in inspector.get_columns(table)}
            missing_cols = set(TABLE_COLUMNS[table]) - database_columns
            if missing_cols:
                raise RuntimeError(f"{table} missing SQL columns: {missing_cols}")
            database_pk = inspector.get_pk_constraint(table).get("constrained_columns") or []
            if set(database_pk) != set(PRIMARY_KEYS[table]):
                raise RuntimeError(f"{table} PRIMARY KEY mismatch: {database_pk}")
            storage = conn.execute(text("""
                SELECT ENGINE FROM information_schema.TABLES
                WHERE TABLE_SCHEMA=:db AND TABLE_NAME=:table
            """), {"db": DB_NAME, "table": table}).scalar_one()
            if str(storage).upper() != "INNODB":
                raise RuntimeError(f"{table}: InnoDB required; got {storage}")
        row = conn.execute(text("SHOW VARIABLES LIKE 'local_infile'")).first()
        local_infile = bool(row and str(row[1]).upper() == "ON")
        print(f"Database verified: {DB_NAME} at {DB_HOST}:{DB_PORT}; LOCAL INFILE={local_infile}")
        return local_infile

# Verify the committed rows really match the expected source prefix.
# For large tables, checks a spread of PKs, not every value/row.
def verified_offset(table, frame):
    with engine.connect() as conn:
        count = int(conn.execute(text(f"SELECT COUNT(*) FROM `{table}`")).scalar_one())
        if count > len(frame):
            raise RuntimeError(f"{table}: DB has {count} rows; expected {len(frame)}")
        if count == 0:
            return 0
        if table == "inventory":
            db_min, db_max = conn.execute(text(
                "SELECT MIN(inventory_id), MAX(inventory_id) FROM inventory"
            )).one()
            ids = frame.iloc[:count]["inventory_id"].astype(int)
            if (ids.nunique() != count or int(ids.max()) - int(ids.min()) + 1 != count
                or int(db_min) != int(ids.min()) or int(db_max) != int(ids.max())):
                raise RuntimeError("Inventory IDs do not match expected contiguous prefix")
            return count
        indexes = sorted({0, min(1, count - 1), count // 8, count // 4,
                          count // 2, count * 3 // 4, count * 7 // 8, count - 1})
        keys = PRIMARY_KEYS[table]
        terms = " AND ".join(f"`{col}` = :key{i}" for i, col in enumerate(keys))
        query = text(f"SELECT 1 FROM `{table}` WHERE {terms} LIMIT 1")
        for idx in indexes:
            row = frame.iloc[idx]
            values = {f"key{i}": row[col] for i, col in enumerate(keys)}
            if conn.execute(query, values).scalar_one_or_none() != 1:
                raise RuntimeError(f"{table}: previously inserted rows do not match source prefix at {idx}")
        return count

# MariaDB TSV with backslash-escaped special characters and \N for SQL NULL.
def encode_tsv(value):
    if value is None or pd.isna(value):
        return r"\N"
    if isinstance(value, (datetime, pd.Timestamp)):
        value = value.strftime("%Y-%m-%d %H:%M:%S")
    elif isinstance(value, date):
        value = value.isoformat()
    return (str(value).replace("\\", "\\\\").replace("\t", "\\t")
            .replace("\n", "\\n").replace("\r", "\\r")
            .replace("\x00", "\\0").replace("\x1a", "\\Z"))

def build_tsv(frame, offset, limit, byte_cap):
    payload = []
    size = 0
    for values in frame.iloc[offset:offset + limit].itertuples(index=False, name=None):
        line = ("\t".join(encode_tsv(v) for v in values) + "\n").encode("utf-8")
        if len(line) > byte_cap:
            raise RuntimeError(f"One row exceeds byte limit ({len(line)} bytes)")
        if payload and size + len(line) > byte_cap:
            break
        payload.append(line)
        size += len(line)
    if not payload:
        raise RuntimeError("No TSV rows created")
    return b"".join(payload), len(payload)

def sql_records(frame, start, end):
    result = []
    for record in frame.iloc[start:end].to_dict("records"):
        clean = {}
        for col, val in record.items():
            if val is None or pd.isna(val):
                clean[col] = None
            elif isinstance(val, (pd.Timestamp, datetime)):
                clean[col] = val.to_pydatetime() if isinstance(val, pd.Timestamp) else val
            elif hasattr(val, "item"):
                clean[col] = val.item()
            else:
                clean[col] = val
        result.append(clean)
    return result

def execute_batch(table, frame, offset, row_cap, byte_cap, use_bulk):
    cols = ", ".join(f"`{c}`" for c in frame.columns)
    if use_bulk:
        payload, n = build_tsv(frame, offset, row_cap, byte_cap)
        path = None
        try:
            with tempfile.NamedTemporaryFile(mode="wb", suffix=".tsv", delete=False) as output:
                path = output.name
                output.write(payload)
            statement = text(rf"""
                LOAD DATA LOCAL INFILE :path
                INTO TABLE `{table}` CHARACTER SET utf8mb4
                FIELDS TERMINATED BY '\t' ESCAPED BY '\\'
                LINES TERMINATED BY '\n'
                ({cols})
            """)
            with engine.begin() as conn:
                outcome = conn.execute(statement, {"path": path})
                warnings = int(conn.execute(text("SHOW COUNT(*) WARNINGS")).scalar_one())
                if warnings:
                    examples = conn.execute(text("SHOW WARNINGS LIMIT 10")).fetchall()
                    raise RuntimeError(f"{table}: LOAD DATA warnings; rolled back: {examples}")
                if outcome.rowcount >= 0 and outcome.rowcount != n:
                    raise RuntimeError(f"{table}: attempted {n}, inserted {outcome.rowcount}")
            return n
        finally:
            if path is not None:
                Path(path).unlink(missing_ok=True)
    end = min(offset + row_cap, len(frame))
    records = sql_records(frame, offset, end)
    placeholders = ", ".join(f":{c}" for c in frame.columns)
    statement = text(f"INSERT INTO `{table}` ({cols}) VALUES ({placeholders})")
    with engine.begin() as conn:
        outcome = conn.execute(statement, records)
        warnings = int(conn.execute(text("SHOW COUNT(*) WARNINGS")).scalar_one())
        if warnings:
            examples = conn.execute(text("SHOW WARNINGS LIMIT 10")).fetchall()
            raise RuntimeError(f"{table}: INSERT warnings; rolled back: {examples}")
        if outcome.rowcount >= 0 and outcome.rowcount != len(records):
            raise RuntimeError(f"{table}: attempted {len(records)}, inserted {outcome.rowcount}")
    return len(records)

# ============================================================
# 5. DRY RUN OR RESUMABLE BULK UPLOAD
# ============================================================
try:
    bulk_possible = check_destination() and PREFER_BULK
    print("Upload method:", "BULK TSV" if bulk_possible else "SQL batch INSERT")
    print("Expected table counts:")
    for table in LOAD_ORDER:
        print(f"  {table:12s} {len(prepared[table]):>10,}")
    print("Currently committed:")
    for table in LOAD_ORDER:
        n = verified_offset(table, prepared[table])
        print(f"  {table:12s} {n:>10,} / {len(prepared[table]):,}")

    if DRY_RUN:
        print("\nDRY RUN PASSED: No records inserted. Set DRY_RUN=False and rerun this cell.")
    else:
        started = time.perf_counter()
        total_new = 0
        stop_early = False
        for table in LOAD_ORDER:
            frame = prepared[table]
            offset = verified_offset(table, frame)
            row_cap = BULK_ROWS if bulk_possible else SQL_ROWS
            byte_cap = BULK_BYTES
            fail_count = 0
            report_offset = offset
            print(f"\n{table}: resuming at {offset:,}/{len(frame):,}", flush=True)

            while offset < len(frame):
                budget = None if MAX_NEW_ROWS_PER_RUN is None else MAX_NEW_ROWS_PER_RUN - total_new
                if budget is not None and budget <= 0:
                    stop_early = True
                    break
                rows_this_batch = min(row_cap, len(frame) - offset)
                if budget is not None:
                    rows_this_batch = min(rows_this_batch, budget)
                before = offset
                try:
                    n = execute_batch(table, frame, offset, rows_this_batch, byte_cap, bulk_possible)
                    offset += n
                    total_new += n
                    fail_count = 0
                    if offset - report_offset >= PROGRESS_EVERY or offset == len(frame):
                        print(f"{table}: {offset:,}/{len(frame):,} committed | "
                              f"new this run: {total_new:,} | "
                              f"elapsed {time.perf_counter()-started:.1f}s", flush=True)
                        report_offset = offset
                except (OperationalError, InterfaceError) as exc:
                    fail_count += 1
                    print(f"{table}: connection error at offset {before:,}: {str(exc)[:170]}", flush=True)
                    engine.dispose()
                    if fail_count > MAX_RETRIES:
                        raise RuntimeError(f"{table}: repeated connection errors; rerun unchanged to resume") from exc
                    row_cap = max(50, row_cap // 2)
                    byte_cap = max(32_768, byte_cap // 2)
                    time.sleep(min(2 ** fail_count, 20))
                    # Commit acknowledgement might have been lost: verify DB before retry.
                    confirmed = verified_offset(table, frame)
                    if not (before <= confirmed <= before + rows_this_batch):
                        raise RuntimeError(f"{table}: unexpected offset {before} -> {confirmed}; concurrent write?")
                    total_new += confirmed - before
                    offset = confirmed
                    print(f"{table}: resumed at {offset:,}; retry chunk <= {row_cap:,}", flush=True)

            if stop_early:
                print(f"\nBENCHMARK STOP after inserting {total_new:,} rows in this run")
                break
            actual = verified_offset(table, frame)
            if actual != len(frame):
                raise RuntimeError(f"{table}: not complete ({actual:,}/{len(frame):,})")
            print(f"{table}: COMPLETE ({actual:,})", flush=True)

        if not stop_early:
            print("\nALL 7 TABLES UPLOADED. Confirmed row counts:")
            for table in LOAD_ORDER:
                print(f"  {table:12s} {verified_offset(table, prepared[table]):>10,}")
            print(f"Total new records this run: {total_new:,}")
            print(f"Elapsed time: {time.perf_counter()-started:.1f}s")
finally:
    engine.dispose()
```
