### Final Data

```python
df_final = df[['order_id', 'order_date',  'order_year', 'order_month', 'order_day',
       'order_hour', 'order_minute', 'order_second','is_weekend',
       'order_status', 'customer_id', 'customer_name',
       'gender', 'age', 'customer_segment', 'country', 'city',
       'customer_loyalty_score', 'total_orders_by_customer',
       'account_creation_date', 'product_id', 'product_name', 'category',
       'sub_category', 'brand', 'product_rating_avg', 'product_reviews_count',
       'stock_quantity', 'unit_price_usd', 'quantity', 'discount_percent',
        'total_price_usd', 'cost_usd', 'profit_usd',
       'tax_usd', 'payment_method', 'payment_status',
       'installment_plan', 'shipping_method', 'shipping_cost_usd',
       'delivery_days', 'shipping_country', 'warehouse_location',
       'delivery_status', 'device_type'
        ]]
```

```python
df_final.isna().sum()
```

```python
df_final['product_id'] = pd.factorize(df_final['product_name'])[0] + 1
```

```python
df_final["customer_id"] = df_final['customer_id'].astype(str) + '' + df_final.groupby('customer_id')['customer_name'].transform(lambda x: pd.factorize(x)[0]).astype(str)
```

```python
df_final['order_id'] = df_final['order_id'].astype(str) + '' + df_final.groupby('order_id').cumcount().astype(str)
```
